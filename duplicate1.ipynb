```python
from typing import List
```

```python
def hasDuplicate(nums: List[int]) -> bool:
    empty_list = []
    for num in nums:
        if num in empty_list:
            return True
        empty_list.append(num)
    return False 
```

```python
print(hasDuplicate(array))
```
