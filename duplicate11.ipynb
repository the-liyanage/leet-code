```python
from typing import List
```

```python


class Solution:
    def hasDuplicates(self, num:List[int]) -> bool:
        empty_set = set()
        for number in num:
            if number in empty_set:
                return True
            empty_set.add(number)
        return False
```

```python
nums = [1, 2, 3, 4, 4, 5]
sol = Solution()
```

```python
print(sol.hasDuplicates(nums))
```
