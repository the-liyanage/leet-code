```python
groups = ["hat", "tah", "rat", ]
letter_count = {}
for word in groups:
    letter_count[word] = letter_count.get(word, 0 ) +1
print(letter_count)  
```

```python
groups = ["hat", "tah", "rat", ]
all_count= {}

for word in groups:
    letter_count= {}
    for letter in word:
        letter_count[letter] = letter_count.get(letter, 0) + 1
    all_count[word] = letter_count
print(all_count)
```
