```python
nums = [10, 2, 7,4]
target = 11
```

```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        empty_map = {}
        for i, n in enumerate(nums):
            diff = target - n
            if diff in empty_map:
                return [empty_map[diff], i]
            empty_map[n] = i


sol = Solution()
print(sol.twoSum(nums, target))
```
